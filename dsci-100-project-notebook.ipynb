#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
basketball_metrics <- womens_basketball_tidy |>
    mutate(max_wins = max(tourney_wins)) |> #creating a column that shows the maximum tourney wins
    mutate(max_losses = max(tourney_losses)) |> #creating a column that shows the maximum tourney losses
    mutate(min_wins = min(tourney_wins)) |>  #creating a column that shows the minimum tourney wins
    mutate(min_losses = min(tourney_losses)) |>  #creating a column that shows the minimum tourney losses
    mutate(mean_wins = mean(tourney_wins)) |> #creating a column that shows the mean tourney wins
    mutate(mean_losses = mean(tourney_losses)) |> #creating a column that shows the mean tourney losses
    select(max_wins, max_losses, min_wins, min_losses, mean_wins, mean_losses) |> #selecting only the rows with the summarized values
    slice(1) #showing only one row

basketball_metrics
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).

```r
basketball_na <- womens_basketball_tidy |>
   mutate(tourney_wins = sum(is.na (tourney_wins))) |> # using is.na() to find all NA values and then counting them with sum(), turning this value into a new column with mutate
   mutate(tourney_losses = sum(is.na (tourney_losses))) |> # using is.na() to find all NA values and then counting them with sum(), turning this value into a new column with mutate 
   mutate(tourney_finish = sum(is.na (tourney_finish))) |> # using is.na() to find all NA values and then counting them with sum(), turning this value into a new column with mutate 
   slice(1)
basketball_na
```

Visualization 1: This visualization looks at the relationship between the number of tournament wins and tournament finishing position. This is relevant to the analysis because tourney wins is one of the predictors I will use in the classification analysis, so understanding the relationship between these two variables is important. From this visualization we can see that having the most number of wins does not necessarily mean a higher finishing position. 

```r
options(repr.plot.width = 10, repr.plot.height = 7)

wins_vs_tourney_finish <- womens_basketball_tidy |>
    ggplot(aes (x = reorder(tourney_finish, tourney_wins), y = tourney_wins)) + #set tourney finish as x axis and tourney wins as y-axis
    geom_bar(stat = "identity") + #ensure that bar height is reflecting of the number of times a variable appears in dataset
    labs(x = "Tournament Finish Position", y = "Number of Games Won") + #giving axes names
    ggtitle("Final Tournament Position vs Number of Games Won") #giving the plot a title
wins_vs_tourney_finish
```

Visualization 2: This visualization explores the relationship between the other predictor, tournament losses and tournament finish position. It is relevant because it allows us to understand the relationship between the other predictor and the category we are classifying in the analysis. Here we can see that having a low number of losses seems to result in a higher finishing position. 

```r
options(repr.plot.width = 12, repr.plot.height = 8)

losses_vs_tourney_finish <- womens_basketball_tidy |>
    ggplot(aes (x = reorder(tourney_finish, tourney_losses), y = tourney_losses)) + #set tourney finish as x axis and tourney wins as y-axis
    geom_bar(stat = "identity") + #ensure that bar height is reflecting of the number of times a variable appears in dataset
    labs(x = "Tournament Finish Position", y = "Number of Games Lost") + #giving axes names
    ggtitle("Final Tournament Position vs Number of Games Lost")#giving the plot a title
losses_vs_tourney_finish
```
